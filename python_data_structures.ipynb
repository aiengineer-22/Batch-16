# Python Data Structures


Python has four main built-in containers:

| Structure | Ordered? | Changeable? | Duplicates allowed? | Homogeneous or Heterogeneous? |
|---|---|---|---|---|
| **List** `[]` | Yes | Yes (mutable) | Yes | Both — can hold same type (`[1, 2, 3]`) or mixed types (`[1, "eggs", True]`) |
| **Tuple** `()` | Yes | No (immutable) | Yes | Both — same type (`(1, 2, 3)`) or mixed (`("Hafiza", 25, "Karachi")`) |
| **Dictionary** `{}` | Yes (insertion order) | Yes (mutable) | Keys: No. Values: Yes | Both — keys and values can each be same type or mixed types |
| **Set** `{}` | No | Yes (mutable) | No | Both — but every item must be an immutable type (numbers, strings, tuples — not lists or dicts) |

# 1. Lists

A **list** is an ordered, changeable collection.

A list is like a **to-do list on paper** — you can cross items off, add new ones at the bottom or squeeze one in the middle, reorder the whole thing, or copy it onto a new page.

````python
# Creating a list
# name_of_list = [item1, item2]
groceries = ["milk", "eggs", "bread", "eggs"]
#index         0       1        2        3
print(groceries)
print(type(groceries))

# Duplicates are allowed (notice "eggs" appears twice) and order is preserved
````

## **List Methods**

## 1.1 `append()` — add ONE item to the end

````python
groceries = ["milk", "eggs", "bread"]

# Use case: user adds a new item to their shopping list
groceries.append("butter")
print(groceries)

# append() ALWAYS adds exactly one item, even if that item is itself a list
groceries.append(["cheese", "yogurt"])
print(groceries)   # notice the last element is a nested list, not two new items

groceries.append(('puck', 'almarai'))
print(groceries)
````

## 1.2 `extend()` — add MULTIPLE items to the end

````python
groceries = ["milk", "eggs", "bread"]

groceries.extend(['facewash', 'body wash'])
print(groceries)

# Use case: merging a second list of items into the first, item by item
more_items = ["cheese", "yogurt"]
groceries.extend(more_items)
print(groceries)   # each item added separately, unlike append()

# extend() also works with a string — it adds each CHARACTER separately
letters = ["a", "b"]
letters.extend("cd")
print(letters)
````

## 1.3 `insert()` — add an item at a SPECIFIC position

````python
groceries = ["milk", "eggs", "bread"]

# insert(index, item)

# Use case: you forgot "sugar" and want it at the very start of the list
groceries.insert(0, "sugar")
print(groceries)
````

## 1.4 `remove()` — delete an item BY VALUE

````python
groceries = ["milk", "eggs", "bread", "eggs"]
print(groceries)

# Use case: you bought the milk, remove it from the list
groceries.remove("milk")
print(groceries)

# If the value appears more than once, remove() deletes only the FIRST match
groceries.remove("eggs")
print(groceries)   # only one "eggs" removed, the second one stays

# If the value doesn't exist, Python raises an error - try it:
# groceries.remove("chocolate")   # uncomment to see the ValueError
````

## 1.5 `pop()` — remove an item BY POSITION (and get it back)

````python
groceries = ["milk", "eggs", "bread"]

# Use case: process the LAST item and remove it at the same time (common in stacks)
last_item = groceries.pop()
print("Removed:", last_item)
print("Remaining:", groceries)

# pop() can also take an index to remove from a specific spot
groceries = ["milk", "eggs", "bread"]
first_item = groceries.pop(0)
print("Removed:", first_item)
print("Remaining:", groceries)
````

## 1.6 `clear()` — empty the entire list

````python
groceries = ["milk", "eggs", "bread"]

# Use case: user hits "clear cart" — the list itself stays, just becomes empty
groceries.clear()
print(groceries)
print(type(groceries))   # still a list, just empty
````

## 1.7 `index()` — find the POSITION of a value

````python
groceries = ["milk", "eggs", "bread", "eggs"]

# Use case: find where "bread" sits so you can jump to it
position = groceries.index("bread")
print(position)

# If a value repeats, index() returns the FIRST occurrence
print(groceries.index("eggs"))

# You can also search within a range: index(value, start, end)
print(groceries.index("eggs", 2))   # search starting from position 2 onward
````

## 1.8 `count()` — how many times does a value appear?

````python
groceries = ["milk", "eggs", "bread", "eggs", "eggs"]

# Use case: check how many duplicate entries exist before cleaning the list
print(groceries.count("eggs"))
print(groceries.count("milk"))
print(groceries.count("chocolate"))   # 0, not an error, if it's not found
````

## 1.9 `sort()` — reorder the list IN PLACE

````python
numbers = [5, 2, 9, 1, 7]
print(numbers)

# Use case: sort exam scores from lowest to highest
numbers.sort()
print(numbers)

# sort() changes the ORIGINAL list and returns nothing (None)
result = numbers.sort()
print(result)   # None -- a very common beginner mistake is writing x = list.sort()

# Sort in reverse (highest to lowest)
numbers.sort(reverse=True)
print(numbers)

# Sorting works on strings too (alphabetical order)
names = ["Zara", "amina", "Bilal"]
names.sort()
print(names)   # uppercase letters sort before lowercase in Python by default
````

## 1.10 `reverse()` — flip the order IN PLACE

````python
queue = ["Ali", "Sara", "Bilal"]

# Use case: reverse a queue so the last person becomes first
queue.reverse()
print(queue)

# Note: reverse() just flips the current order, it does NOT sort
````

## 1.11 `copy()` — make a real, independent copy

````python
original = ["milk", "eggs", "bread"]

# The WRONG way -- this does NOT copy, it just gives the same list a second name
same_list = original
same_list.append("butter")
print(original)   # original changed too! Both names point to the SAME list in memory

# The RIGHT way -- copy() creates a separate, independent list
original = ["milk", "eggs", "bread"]
backup = original.copy()
backup.append("butter")
print("original:", original)   # unaffected
print("backup:", backup)       # only this one changed
````

````python
# Indexing and slicing
languages = ['C++', 'C#', 'Java', 'Python', 'Ruby', 'Go']
print(languages[0])
print(languages[2:4])
print(languages[-1])
````

## 1.12 List — quick recap table

| Method | What it does | Changes list? | Returns |
|---|---|---|---|
| `append(x)` | add one item at the end | Yes | `None` |
| `extend(iterable)` | add many items at the end | Yes | `None` |
| `insert(i, x)` | add item at position `i` | Yes | `None` |
| `remove(x)` | delete first match of value `x` | Yes | `None` |

| `reverse()` | flip current order | Yes | `None` |
| `copy()` | independent duplicate | No | a new list |

````python
# Track daily expenses and use `sum()` with a loop to total them
expenses = []

while True:
    expense = float(input("Enter an expense (0 to stop): "))

    if expense == 0:
        break

    expenses.append(expense)
    print(expenses)

total = sum(expenses)

print("Expenses:", expenses)
print("Total expenses:", total)
````

## 1.13 List comprehension — a shortcut for building a list with a loop



```python
squares = []
for n in range(1, 6):
    squares.append(n ** 2)
```

**List comprehension** does the exact same thing, in one line, using this syntax:
 - new_list = [expression for item in iterable]


Read it left to right like this: *"put `expression` into `new_list`, for each `item` in `iterable`"*.

| Part | What it means | Matches which part of the loop version |
|---|---|---|
| `expression` | what to actually store in the new list | `n ** 2` |
| `item` | the variable name for each element as you go | `n` |
| `iterable` | what you're looping over | `range(1, 6)` |

You can also add a condition at the end to filter which items get included:
 - new_list = [expression for item in iterable if condition]


This only adds `expression` to the list when `condition` is `True` — anything that fails the condition is simply skipped.

````python
# Basic loop version
squares = []
for n in range(1, 6):
    squares.append(n ** 2)
print(squares)

# Same result with list comprehension
squares = [n ** 2 for n in range(1, 6)]
print(squares)

# With a condition -- only keep even numbers
evens = [n for n in range(1, 11) if n % 2 == 0]
print(evens)

# Comprehension over a list of strings -- get the length of each word
words = ["apple", "fig", "banana", "kiwi"]
lengths = [len(word) for word in words]
print(lengths)

# Comprehension with a condition -- keep only words longer than 3 letters
long_words = [word for word in words if len(word) > 3]
print(long_words)

# Transform every item -- convert a list of names to uppercase
names = ["ali", "sara", "bilal"]
upper_names = [name.upper() for name in names]
print(upper_names)
````

````python
even_odd = ['even' if n%2 ==0 else 'odd' for n in range(1,21)]
print(even_odd)
````

# 2. Tuples

A **tuple** is **immutable** — once created, it cannot be changed. No adding, no removing, no reordering.

Analogy: a tuple is like a **birth certificate** or a **GPS coordinate** — (latitude, longitude) shouldn't suddenly change because you looked at it. Because tuples can't change, Python gives them far fewer methods than lists — only two.

````python
# Creating a tuple
# name_of_tuple = (value1, value2)
person = ("Ali", 20, "Karachi")   # (name, age, city)
print(person)
print(type(person))

# Indexing and slicing work exactly like lists
print(person[0])
print(person[0:1])

# But trying to change it raises an error:
# person[0] = "Sara"   # uncomment to see: TypeError: 'tuple' object does not support item assignment
````

````python
# Creating a tuple
coordinates = (24.8607, 67.0011)   # (latitude, longitude) of Karachi
print(coordinates)
print(type(coordinates))

# Indexing and slicing
print(coordinates[0])
print(coordinates[0:1])

# But trying to change it raises an error:
# coordinates[0] = 25.0   # uncomment to see: TypeError: 'tuple' object does not support item assignment
````

## 2.1 `count()` — how many times does a value appear?

````python
student = ("Ali", 21, "Ali", "Computer Science")

# Use case: check how many times a value repeats in a fixed record
print(student.count("Ali"))
print(student.count("Sara"))   # 0 if not found
````

## 2.2 `index()` — find the POSITION of a value

````python
days = ("Mon", "Tue", "Wed", "Thu", "Fri")

# Use case: find which position "Wed" sits at in a fixed weekly schedule
print(days.index("Wed"))

# Same rule as lists: raises an error if the value isn't found
# days.index("Sun")   # uncomment to see the ValueError
````

## 2.3 Why use a tuple?

- **Protection**: data that should never accidentally change (coordinates, RGB colors, a date-of-birth record)
- **Speed**: tuples are slightly faster than lists because Python knows they'll never resize
- **Dictionary keys**: tuples can be used as dictionary keys, lists cannot (you'll see why once we cover dictionaries below — it's because keys must be unchangeable)
- **Multiple return-style values**: when you want to group a few related fixed values together, e.g. `person = ("Hafiza", 25, "Karachi")`

# 3. Dictionaries

A **dictionary** stores data as **key : value** pairs. Instead of finding something by its position (like a list), you find it by its **name** (the key).

Analogy: a phone book — you don't look up a person's number by "the 47th entry", you look them up **by name**, and that name (key) gives you their number (value).

````python
# Creating a dictionary
# name_of_dictionary = {key:value, key:value}
student = {
    "name": "Hafiza",
    "age": 25,
    "city": "Karachi",
    "is_enrolled": True
}
print(student)
print(type(student))

# Access a value using its key (NOT an index number)
print(student["name"])
# print(student["mail"]) will throw an error


# Keys must be unique -- assigning to an existing key OVERWRITES it, doesn't duplicate it
student["age"] = 26
print(student)
````

## 3.1 `get()` — safely fetch a value by key

````python
student = {"name": "Hafiza", "age": 25}

# Use case: look up a key that MIGHT not exist, without crashing the program
print(student.get("name"))
print(student.get("email"))          # not found -> returns None, no error
print(student["email"] if "email" in student else "N/A")   # compare to the crash-prone way

# get() lets you set a custom fallback/default value
print(student.get("email", "no email on file"))

# Compare to plain [] access, which CRASHES if the key is missing:
# print(student["email"])   # uncomment to see the KeyError
````

## 3.2 `keys()` — get all the keys

````python
student = {"name": "Hafiza", "age": 25, "city": "Karachi"}

print(student.keys())

for key in student.keys():
    print(key)

# Handy with "in" to check if a key exists anywhere in the dictionary
print("email" in student.keys())
````

## 3.3 `values()` — get all the values

````python
student = {"name": "Hafiza", "age": 25, "city": "Karachi"}

# Use case: loop through just the DATA, ignoring what each field is called
for value in student.values():
    print(value)

# Handy with "in" to check if a value exists anywhere in the dictionary
print("Karachi" in student.values())
print("Lahore" in student.values())
````

## 3.4 `items()` — get key-value pairs together

````python
student = {"name": "Hafiza", "age": 25, "city": "Karachi"}

print(student.items())

# Use case: print a full readable record, field by field
for key, value in student.items():
    print(key, "->", value)

# This is the most common way to loop over a dictionary in real code
````

## 3.5 `update()` — add or change multiple key-value pairs at once

````python
student = {"name": "Hafiza", "age": 25}

# Use case: a form submits several updated fields at once
student.update({"age": 26, "city": "Karachi"})
print(student)   # "age" was overwritten, "city" was added

# You can also update from another dictionary variable
new_info = {"is_enrolled": True}
student.update(new_info)
print(student)
````

## 3.6 `pop()` — remove a key and get its value back

````python
student = {"name": "Hafiza", "age": 25, "city": "Karachi"}

# Use case: remove a field you no longer need, but capture what it was
removed_value = student.pop("city")
print("Removed:", removed_value)
print("Remaining:", student)

# pop() lets you give a default so it doesn't crash if the key is missing
print(student.pop("email", "key was not there"))
````

## 3.7 `popitem()` — remove the LAST inserted key-value pair

````python
student = {"name": "Hafiza", "age": 25, "city": "Karachi"}

# Use case: undo the most recent addition to the dictionary
last_pair = student.popitem()
print("Removed:", last_pair)   # returns a tuple: (key, value)
print("Remaining:", student)
````

## 3.8 `setdefault()` — get a value, or insert it if missing

````python
student = {"name": "Hafiza", "age": 25}

# Use case: fill in a field only if it doesn't already exist
student.setdefault("city", "Karachi")
print(student)   # "city" added because it wasn't there

# If the key ALREADY exists, setdefault does nothing and just returns the current value
student.setdefault("name", "Someone Else")
print(student)   # "name" is unchanged, still "Hafiza"
student.setdefault()
````

## 3.9 `clear()` — empty the dictionary

````python
student = {"name": "Hafiza", "age": 25}

student.clear()
print(student)
print(type(student))   # still a dict, just empty
````

## 3.10 `copy()` — make an independent copy

````python
original = {"name": "Hafiza", "age": 25}

# Same trap as lists: this does NOT copy, both names point to the same dictionary
same_dict = original
same_dict["age"] = 99
print(original)   # changed too!

# The correct way:
original = {"name": "Hafiza", "age": 25}
backup = original.copy()
backup["age"] = 99
print("original:", original)
print("backup:", backup)
````

## 3.11 `fromkeys()` — build a new dictionary from a list of keys

````python
# Use case: initialize a set of fields with the SAME starting value
fields = ["name", "age", "city"]
blank_form = dict.fromkeys(fields, "")
print(blank_form)

# If you don't give a default value, it fills with None
scores = dict.fromkeys(["Ali", "Sara", "Bilal"])
print(scores)
````

## 3.12 Dictionary — quick recap table

| Method | What it does | Changes dict? | Returns |
|---|---|---|---|
| `get(key, default)` | safely fetch a value | No | value or default |
| `keys()` | all keys | No | view of keys |
| `values()` | all values | No | view of values |
| `items()` | all key-value pairs | No | view of (key, value) tuples |
| `update(other)` | add/overwrite many pairs | Yes | `None` |
| `pop(key, default)` | remove one key, return its value | Yes | the removed value |
| `popitem()` | remove the last-inserted pair | Yes | (key, value) tuple |
| `setdefault(key, default)` | get value, insert if missing | Sometimes | the value |
| `clear()` | remove everything | Yes | `None` |
| `copy()` | independent duplicate | No | a new dict |
| `dict.fromkeys(keys, value)` | build a new dict from a list of keys | No | a new dict |

# 4. Sets

A **set** is an unordered collection of **unique** items — duplicates are automatically thrown out, and there's no guaranteed order or index.


````python
# Creating a set
# name_of_Set = {item1, item2}
fruits = {"apple", "banana", "apple", "mango"}
print(fruits)   # duplicate "apple" is automatically removed
print(type(fruits))

# Sets have NO indexing -- this will fail:
# print(fruits[0])   # uncomment to see: TypeError: 'set' object is not subscriptable

# An empty {} is a DICTIONARY, not a set. To make an empty set you must write:
empty_set = set()
print(type(empty_set))
````

## 4.1 `add()` — add ONE item

````python
fruits = {"apple", "banana"}

# Use case: add a new unique tag to a collection
fruits.add("mango")
print(fruits)

# Adding a value that already exists does nothing (no error, no duplicate)
fruits.add("apple")
print(fruits)
````

## 4.2 `remove()` — delete an item (error if missing)

````python
fruits = {"apple", "banana", "mango"}

fruits.remove("banana")
print(fruits)

# If the item isn't in the set, remove() raises an error
# fruits.remove("grape")   # uncomment to see the KeyError
````

## 4.3 `discard()` — delete an item (NO error if missing)

````python
fruits = {"apple", "banana", "mango"}

# Use case: remove an item when you're not sure it's even there
fruits.discard("banana")
print(fruits)

fruits.discard("grape")   # does nothing, no crash -- this is the key difference from remove()
print(fruits)
````

## 4.4 `pop()` — remove and return a RANDOM item

````python

fruits = {"apple", "banana", "mango"}

# Use case: pull out "some" item when order genuinely doesn't matter
removed = fruits.pop()
print("Removed:", removed)
print("Remaining:", fruits)

# You can't control WHICH item gets removed -- sets have no order, so there's no "first" or "last"
````

## 4.5 `clear()` — empty the set

````python
fruits = {"apple", "banana", "mango"}
fruits.clear()
print(fruits)
````

## 4.6 `copy()` — make an independent copy

````python
original = {"apple", "banana"}
backup = original.copy()
backup.add("mango")
print("original:", original)
print("backup:", backup)
````

## 4.7 Set math — this is where sets really shine
Sets come from mathematics, so they support the same operations as Venn diagrams: union, intersection, difference. Every one of these has TWO versions:
- one that **returns a new set** and leaves the originals untouched
- one ending in `_update` that changes the set **in place**

````python
class_a = {"Ali", "Sara", "Bilal", "Hina"}
class_b = {"Sara", "Hina", "Zain", "Omar"}

# UNION -- everyone in EITHER class (combines both, no duplicates)
print(class_a.union(class_b))

# INTERSECTION -- students in BOTH classes
print(class_a.intersection(class_b))

# DIFFERENCE -- in class_a but NOT in class_b
print(class_a.difference(class_b))

# SYMMETRIC DIFFERENCE -- in exactly ONE of the two classes, not both
print(class_a.symmetric_difference(class_b))
````

## 4.8 `update()`, `intersection_update()`, `difference_update()`, `symmetric_difference_update()`

````python
# These do the SAME math as above, but change the set IN PLACE instead of returning a new one

class_a = {"Ali", "Sara", "Bilal", "Hina"}
class_b = {"Sara", "Hina", "Zain", "Omar"}

# update() -- add everything from class_b into class_a directly (like union, but in place)
a_copy = class_a.copy()
a_copy.update(class_b)
print("update:", a_copy)

# intersection_update() -- keep only what's shared with class_b
a_copy = class_a.copy()
a_copy.intersection_update(class_b)
print("intersection_update:", a_copy)

# difference_update() -- remove anything that's also in class_b
a_copy = class_a.copy()
a_copy.difference_update(class_b)
print("difference_update:", a_copy)

# symmetric_difference_update() -- keep only what's NOT shared
a_copy = class_a.copy()
a_copy.symmetric_difference_update(class_b)
print("symmetric_difference_update:", a_copy)
````

## 4.9 `issubset()`, `issuperset()`, `isdisjoint()` — comparing sets

````python
small_group = {"Ali", "Sara"}
class_a = {"Ali", "Sara", "Bilal", "Hina"}
other_class = {"Zain", "Omar"}

# issubset() -- is EVERY item of small_group inside class_a?
print(small_group.issubset(class_a))

# issuperset() -- does class_a CONTAIN all of small_group?
print(class_a.issuperset(small_group))

# isdisjoint() -- do the two sets share NO items at all?
print(class_a.isdisjoint(other_class))
print(class_a.isdisjoint(small_group))
````

## 4.10 Set — quick recap table

| Method | What it does | Changes set? | Returns |
|---|---|---|---|
| `add(x)` | add one item | Yes | `None` |
| `remove(x)` | delete item (error if missing) | Yes | `None` |
| `discard(x)` | delete item (safe if missing) | Yes | `None` |
| `pop()` | remove a random item | Yes | the removed item |
| `clear()` | remove everything | Yes | `None` |
| `copy()` | independent duplicate | No | a new set |
| `union(other)` | items in either set | No | a new set |
| `intersection(other)` | items in both sets | No | a new set |
| `difference(other)` | items only in this set | No | a new set |
| `symmetric_difference(other)` | items in exactly one set | No | a new set |
| `update(other)` | union, but in place | Yes | `None` |
| `intersection_update(other)` | intersection, but in place | Yes | `None` |
| `difference_update(other)` | difference, but in place | Yes | `None` |
| `symmetric_difference_update(other)` | symmetric difference, but in place | Yes | `None` |
| `issubset(other)` | is this set fully inside `other`? | No | `True`/`False` |
| `issuperset(other)` | does this set fully contain `other`? | No | `True`/`False` |
| `isdisjoint(other)` | no shared items at all? | No | `True`/`False` |

# 5. Putting it all together — which one do I use?

Ask yourself these questions, in order:

1. **Do I need to look things up by a name/label instead of a position?** -> use a **dictionary**
2. **Do I need to guarantee every item is unique, and don't care about order?** -> use a **set**
3. **Will this collection of values ever need to change (add/remove/reorder)?**
   - Yes -> use a **list**
   - No, these values are fixed and should stay protected -> use a **tuple**

## Quick decision examples
- A student's full record (name, age, city) that shouldn't change -> **tuple**
- A shopping cart where items get added/removed -> **list**
- A username -> password lookup table -> **dictionary**
- A list of unique tags on a blog post -> **set**
