# Python Operators

---
## What is an operator, really?

An **operator** is a symbol that tells Python to perform an action on one or more values. Those values are called **operands**.

```
5   +   3
↑   ↑   ↑
operand  operator  operand
```

### The families we'll cover
| # | Family | Example | What it answers |
|---|--------|---------|------------------|
| 1 | Arithmetic | `a + b` | "What's the math result?" |
| 2 | Assignment | `a += b` | "How do I store/update a value?" |
| 3 | Comparison | `a > b` | "How do these two compare?" |
| 4 | Logical | `a and b` | "Are these conditions true together?" |
| 5 | Identity | `a is b` | "Are these the *same object* in memory?" |
| 6 | Membership | `a in b` | "Does this exist inside that?" |
| 7 | Ternary | `x if cond else y` | "One-line if/else?" |

# 1. Arithmetic Operators

Arithmetic operators perform mathematical calculations on numbers.

| Operator | Name | Example | Meaning |
|----------|------|---------|---------|
| `+` | Addition | `5 + 3` | Sum |
| `-` | Subtraction | `5 - 3` | Difference |
| `*` | Multiplication | `5 * 3` | Product |
| `/` | True Division | `5 / 3` | Division — **always returns a float** |
| `//` | Floor Division | `5 // 3` | Whole groups only (rounds down toward -inf) |
| `%` | Modulus | `5 % 3` | The **remainder** after division |
| `**` | Exponentiation | `5 ** 3` | Power |

**Decision rule:** Use `/` for an exact decimal answer. Use `//` when you only care about how many whole times something fits. Use `%` when you care about what's left over.

````python
# Basic arithmetic
a = 17
b = 5

print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)   # -> true division, ALWAYS gives a float
print("a // b =", a // b)  # -> floor division, drops the decimal part
print("a % b  =", a % b)
print("a ** b =", a ** b)
````

# 2. Assignment Operators

Augmented assignment/ Compound operators combine an arithmetic operation with assignment in one step.

| Operator | Equivalent to |
|----------|---------------|
| `=` | plain assignment |
| `+=` | `x = x + y` |
| `-=` | `x = x - y` |
| `*=` | `x = x * y` |
| `/=` | `x = x / y` |
| `//=` | `x = x // y` |
| `%=` | `x = x % y` |
| `**=` | `x = x ** y` |

**Decision rule:** Use `+=` etc. whenever you're updating a variable based on its own current value — shorter, and reads as "update".

````python
score = 10
print("start:", score)

score += 5     # same as: score = score + 5
print("after += 5:", score)      # 15

score -= 3
print("after -= 3:", score)      # 12

score *= 2
print("after *= 2:", score)      # 24

score //= 5
print("after //= 5:", score)     # 4

score **= 3
print("after **= 3:", score)     # 64
````

# 3. Comparison (Relational) Operators

Comparison operators compare two values and always return a boolean.

| Operator | Meaning |
|----------|---------|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

**Decision rule:** Use these whenever your code needs to make a decision based on how two values relate — the backbone of every `if` statement.

````python
a, b = 10, 20

print(a == b)   # False
print(a != b)   # True
print(a > b)    # False
print(a < b)    # True
print(a >= 10)  # True -> equal counts as "greater than or equal"
print(b <= 20)  # True
````

# 4. Logical (Boolean) Operators

**Analogy:** Two light switches wired together. `and` is switches in **series** — the light only turns on if BOTH are flipped on. `or` is switches in **parallel** — the light turns on if EITHER is flipped on. `not` flips whatever switch it's given.

**Concept:** Logical operators combine or invert boolean expressions.

| Operator | True when... |
|----------|---------------|
| `and` | **both** operands are truthy |
| `or` | **at least one** operand is truthy |
| `not` | inverts a single operand |

**Decision rule:** Use `and` when every condition must hold, `or` when any one is enough, `not` to flip a condition.

````python
age = 25
has_id = True

can_enter_office = age >= 18 and has_id   # BOTH must be True
print("Can enter office:", can_enter_office)   # True

is_weekend = False
is_holiday = True
day_off = is_weekend or is_holiday      # AT LEAST ONE must be True
print("Day off:", day_off)   # True

print("NOT day_off:", not day_off)   # False
````

# 5. Identity Operators

**Analogy:** Two people can wear identical outfits (equal appearance) but are still two separate humans (different identity). `is` asks the identity question, not the value question.

**Concept:** `is`/`is not` check whether two names point to the **exact same object in memory** — not whether they look equal.

**Decision rule:** Use `is`/`is not` almost exclusively for comparing against `None`, `True`, or `False`. For everything else, use `==`.

````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a == b:", a == b)   # True  -> same VALUES
print("a is b:", a is b)   # False -> different OBJECTS
print("a is c:", a is c)   # True  -> c literally IS the same object as a

print("id(a):", id(a))     # unique memory address
print("id(b):", id(b))     # different from id(a)
print("id(c):", id(c))     # same as id(a)
````

# 6. Membership Operators

**Analogy:** Checking a guest list at a party door: "Is this name ON the list?" You don't care where — just whether it's there at all.

**Concept:** `in`/`not in` test whether a value exists inside a collection (list, tuple, set, dict, string).

**Decision rule:** Use these instead of manually scanning through a collection — clearer to read, and often faster.

````python
name = "Hafiza Arshmah"

print('Hafiza' in name)
print('hafiza' in name)
print('q' in name)
print('m' in name)
````

````python
fruits = ["apple", "banana", "mango"]

print("apple" in fruits)        # True
print("grape" in fruits)        # False
print("grape" not in fruits)    # True

# Works on strings — checks for SUBSTRING presence
sentence = "Python is fun"
print("fun" in sentence)        # True
print("Java" in sentence)       # False

# Works on dictionaries — checks the KEYS by default, not the values
student = {"name": "Arsh", "age": 21}
print("name" in student)          # True  -> checks keys
print("Arsh" in student)           # False -> "Ali" is a VALUE, not a key
print("Arsh" in student.values())  # True  -> explicitly check values instead
````

# 7. The Ternary Operator (Conditional Expression)

**Analogy:** Instead of a full paragraph — "If it's raining, I will bring an umbrella. Otherwise, I will not." — you compress it into one sentence: "I'll bring an umbrella if it's raining, otherwise not."

**Concept:** A one-line `if/else` that produces a value:
```python
value_if_true if condition else value_if_false
```

**Decision rule:** Use this only for short, simple either/or choices. For anything more complex, use a regular `if/elif/else` block.

````python
age = 20

# Long form
if age >= 18:
    status = "adult"
else:
    status = "minor"
print(status)

# Same logic, ternary form
status = "adult" if age >= 18 else "minor"
print(status)
````

# 8. Operator Precedence & Associativity

**Analogy:** "Order of operations" from school math (PEMDAS). Python has the same idea, extended to every operator family above.

**Concept:** **Precedence** decides which operator runs first when several appear together. **Associativity** is the tie-breaker direction (left-to-right or right-to-left) when operators of the *same* precedence appear together.

### Simplified precedence order (top runs first)
| Precedence | Operators | Associativity |
|---|---|---|
| Highest | `()` parentheses | — |
| | `**` exponentiation | **Right-to-left** |
| | `*`, `/`, `//`, `%` | Left-to-right |
| | `+`, `-` | Left-to-right |
| | `<<`, `>>`, `&`, `^`, `\|` | Left-to-right |
| | `==`, `!=`, `<`, `<=`, `>`, `>=`, `is`, `in` | Left-to-right (chained) |
| | `not` | Right-to-left |
| | `and` | Left-to-right |
| Lowest | `or` | Left-to-right |

**Decision rule:** You don't need to memorize the whole table — remember the *shape*: math beats comparisons, which beat `not`, which beats `and`, which beats `or`. **When in doubt, add parentheses** — they cost nothing and remove all ambiguity.

### What "right-associative" means

**Analogy:** Imagine three friends holding hands in a line: A–B–C. Associativity asks: when two operators of the same kind sit back to back, which pair links up first?
- **Left-to-right:** A and B link first, then that result links with C. Like reading a sentence.
- **Right-to-left:** B and C link first, then A links with that result — you start from the right end.

Most operators (`+`, `-`, `*`, `/`) are left-associative — matches how you'd read `10 - 4 - 2` naturally: `(10-4)-2 = 4`. But `**` is the big exception — it's **right-associative**.

````python
# If ** were left-associative (like + or -), it would group like this:
left_way = (2 ** 3) ** 2
print("If left-to-right: (2**3)**2 =", left_way)     # 64

# But ** is actually right-associative, grouping like this instead:
right_way = 2 ** (3 ** 2)
print("Actual right-to-left: 2**(3**2) =", right_way)   # 512

# What Python gives with no parentheses at all:
print("Python result: 2 ** 3 ** 2 =", 2 ** 3 ** 2)      # 512, matches the right-to-left version
````

````python
# More precedence in action
print(2 + 3 * 4)          # 14, not 20 -> * runs before +
print((2 + 3) * 4)        # 20 -> parentheses force + first

print(10 - 4 - 2)         # 4 -> left-to-right: (10 - 4) - 2

print(5 > 3 and 2 < 4)    # True -> comparisons run before 'and'

print(not True and False)     # False -> (not True) and False -> False and False
print(not (True and False))   # True  -> parentheses change the grouping entirely!
````
