# Python Control Structures

---
## What is a "control structure"?

**Analogy:** Think of a fork in the road with a sign: "If the light is green, go straight. Otherwise, stop." You don't drive straight through no matter what — you **check a condition first**, and the condition decides which path you take.

**Concept:** A control structure is code that lets Python **choose** which block of instructions to run, based on whether a condition is `True` or `False`. Every condition you'll write here is built using the operators you already know: `==`, `>`, `and`, `or`, `in`, etc.

### The three shapes we'll cover
| Shape | Looks like | Meaning |
|---|---|---|
| `if` | one path, or nothing | "Do this ONLY if the condition holds" |
| `if / else` | two paths | "Do this if True, otherwise do that" |
| `if / elif / else` | many paths | "Check conditions in order, run whichever first one matches" |
| nested `if` | a path inside a path | "Once inside one condition, check ANOTHER condition" |

# 1. The Plain `if` Statement

**Analogy:** A doorman who only lets you in if you show a ticket. No ticket, no matter — he just doesn't do anything. There's no "otherwise" action; the door either opens or it stays exactly as it was.

**Concept:** An `if` statement runs a block of code **only when its condition is `True`**. If the condition is `False`, Python just skips the block entirely and moves on.

```python
if condition:
    # this block runs ONLY if condition is True
```

**Decision rule:** Use a plain `if` when there's an action to take for one case, and genuinely nothing to do otherwise.

````python
age = 20

if age >= 18:
    print("You are allowed to vote.")

print("Program continues here either way.")
````

````python
age = 15

if age >= 18:
    print("This line is INSIDE the if — only runs when age >= 18")
print("This line is OUTSIDE the if (no indent) — runs no matter what")

# Nothing printed for the first line above, because age is 15.
# The second line still ran, because it's not indented under the if.
````

### Building the condition using operators

````python
temperature = 40
is_weekend = True
basket = ["milk", "eggs", "bread"]

if temperature > 35:                      # comparison operator
    print("It's a hot day.")

if is_weekend:                            # a plain boolean value works as a condition too
    print("No work today.")

if "eggs" in basket:                      # membership operator
    print("Eggs are in the basket.")

if temperature > 30 and is_weekend:       # logical operator combining two conditions
    print("Great day for the pool.")
````

# 2. `if / else` — Two Paths

**Analogy:** A traffic light with only two states as far as you're concerned: green (go) or not-green (stop). You always do exactly ONE of the two things — never both, never neither.

**Concept:** `else` gives Python a fallback action to run when the `if` condition is `False`.

```python
if condition:
    # runs when condition is True
else:
    # runs when condition is False
```

**Decision rule:** Use `if/else` whenever there are exactly two outcomes and you need an action for BOTH of them, not just one.

````python
age = 15

if age >= 18:
    print("You can vote.")
else:
    print("You are too young to vote.")
````

````python
number = 17

if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")
````

# 3. `if / elif / else` — Many Paths

**Analogy:** A grading teacher going down a checklist: "Is the score 90 or above? If so, A. If not, is it 80 or above? If so, B. If not, is it 70 or above?..." She checks conditions **in order**, and stops at the very first one that matches.

**Concept:** `elif` (short for "else if") lets you chain multiple conditions. Python checks them **top to bottom** and runs the block under the **first `True` one it finds** — then skips all the rest, even if they'd also be `True`.

```python
if condition_1:
    # runs if condition_1 is True
elif condition_2:
    # runs if condition_1 was False AND condition_2 is True
elif condition_3:
    # runs if both above were False AND condition_3 is True
else:
    # runs if NONE of the above were True
```

**Decision rule:** Use `elif` chains when you have more than two possible outcomes, and they're mutually exclusive (only one should ever run).

````python
score = 82

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print("Grade:", grade)   # B
````

### Order matters — the FIRST match wins, not the "best" match

This is a very common beginner trap. Watch what happens if the conditions are written in the wrong order.

````python
score = 95

# WRONG order — broadest condition checked first
if score >= 60:
    grade = "D or above (unhelpful!)"
elif score >= 90:
    grade = "A"     # never reached! score >= 60 already matched above
print("Bad order result:", grade)

print()

# CORRECT order — most specific / narrowest condition first
if score >= 90:
    grade = "A"
elif score >= 60:
    grade = "D or above"
print("Good order result:", grade)
````

**Decision rule:** When your `elif` conditions overlap (like score ranges), always write them from most specific/strict to most general — otherwise a later, more precise check may never be reached.

### Building `elif` conditions using operators

````python
day = "Sat"

if day in ["Mon", "Tue", "Wed", "Thu", "Fri"]:      # membership operator
    print("Weekday — go to work")
elif day == "Sat" or day == "Sun":                  # logical operator
    print("Weekend — relax")
else:
    print("Not a valid day")
````

# 4. Nested `if` — A Decision Inside a Decision

**Analogy:** A bouncer at a club doesn't just check one thing. First: "Are you 18+?" If yes, THEN he checks a second, separate thing: "Do you have a ticket?" Each question only gets asked if the previous one passed — it's a decision **inside** a decision.

**Concept:** A nested `if` is simply an `if` (or `if/elif/else`) statement placed **inside** the body of another `if`. The inner condition is only checked at all if the outer condition was `True`.

```python
if outer_condition:
    if inner_condition:
        # runs only if BOTH outer_condition AND inner_condition are True
```

**Decision rule:** Use nesting when the second question genuinely only makes sense *after* the first question is answered — otherwise, a single combined condition with `and` is usually cleaner (see the comparison below).

````python
age = 20
has_ticket = True

if age >= 18:
    print("Age check passed.")
    if has_ticket:
        print("Ticket check passed.")
        print("Entry allowed!")
    else:
        print("Sorry, you need a ticket.")
else:
    print("Sorry, you must be 18 or older.")
````

### Nested vs. combined `and` — when to use which

The nested version above and the single combined-condition version below give the **same result** here. But they're not always interchangeable — nesting lets you print something **different at each stage**, which a single `and` condition can't do.

````python
age = 20
has_ticket = True

# Combined version — works fine when you only care about the FINAL yes/no
if age >= 18 and has_ticket:
    print("Entry allowed!")
else:
    print("Entry denied.")

# But this combined version can't tell you WHICH check failed —
# nesting is what you need when each stage needs its own message.
````

### Nesting more than two levels deep

````python
username = "admin"
password = "secret123"
is_account_active = True

if username == "admin":
    if password == "secret123":
        if is_account_active:
            print("Login successful.")
        else:
            print("Account is disabled.")
    else:
        print("Incorrect password.")
else:
    print("Unknown username.")
````

### Gotcha: deep nesting gets hard to read fast

**Analogy:** Nesting is like Russian nesting dolls — one or two levels is fine, but by the fourth or fifth doll, you've lost track of which doll you're even looking at.

Compare the login example above with a flatter version using chained logical operators — often easier to scan, when you truly only care about the final outcome.

````python
username = "admin"
password = "secret123"
is_account_active = True

if username == "admin" and password == "secret123" and is_account_active:
    print("Login successful.")
else:
    print("Login failed.")   # doesn't tell you WHY it failed — that's the tradeoff
````

**Decision rule:** Nest when you need a *different message or action at each stage*. Flatten with `and`/`or` when you only care about the final combined yes/no. If you find yourself nesting more than 2–3 levels deep, it's usually a sign to simplify the logic.

# 5. Logic Building — Combining Operators Inside Conditions

**Analogy:** A single condition is one lock on a door. Combining conditions with `and`/`or`/`not` is like adding more locks — some doors need EVERY lock opened (`and`), some just need ANY ONE lock opened (`or`), and sometimes you want to flip what counts as "locked" entirely (`not`).

This section is pure practice — building real decision logic by combining the operators you already know, inside `if` conditions. No new syntax here, just applying everything together.

````python
# Example 1: eligibility check combining multiple ANDs
age = 22
has_degree = True
years_experience = 3

if age >= 21 and has_degree and years_experience >= 2:
    print("Eligible for the position.")
else:
    print("Not eligible.")
````

````python
# Example 2: combining AND / OR together — parentheses make intent explicit
has_id = True
is_member = False
is_vip_guest = True

# Entry allowed if: (has an ID) AND (is a member OR is a VIP guest)
if has_id and (is_member or is_vip_guest):
    print("Entry allowed.")
else:
    print("Entry denied.")
````

````python
# Example 3: using 'not' to flip a condition
is_banned = False

if not is_banned:
    print("User can log in.")
else:
    print("User is banned.")
````

````python
# Example 4: combining membership + comparison + logical operators together
role = "editor"
allowed_roles = ["admin", "editor", "moderator"]
account_age_days = 40

if role in allowed_roles and account_age_days >= 30:
    print("Can publish content.")
else:
    print("Cannot publish content yet.")
````

### Gotcha: `and`/`or` precedence inside conditions

Remember from the Operators notebook: `and` binds tighter than `or`. When mixing both, add parentheses to make your INTENT explicit — even when the result would be the same without them, it removes all doubt for anyone reading the code later (including future you).

````python
age = 16
has_parent_consent = True
is_holiday_event = False

# Without parentheses — technically correct, but ambiguous to read at a glance
if age >= 18 or has_parent_consent and is_holiday_event:
    print("Can attend (no parens)")
else:
    print("Cannot attend (no parens)")

# WITH parentheses — same result, but the grouping is now obvious
if age >= 18 or (has_parent_consent and is_holiday_event):
    print("Can attend (with parens)")
else:
    print("Cannot attend (with parens)")
````

### Problem 2: Triangle Validity + Type Classifier
Given three side lengths `a`, `b`, `c`:
1. First check if they can even form a valid triangle — the sum of any two sides must be greater than the third.
2. If valid, classify it: `"Equilateral"` (all sides equal), `"Isosceles"` (exactly two sides equal), or `"Scalene"` (all different).

Test with `a, b, c = 5, 5, 8`.

<details>
<summary><b>▶ Solution</b></summary>

```python
a, b, c = 5, 5, 8

if (a + b > c) and (a + c > b) and (b + c > a):
    if a == b == c:
        print("Equilateral triangle")
    elif a == b or b == c or a == c:
        print("Isosceles triangle")
    else:
        print("Scalene triangle")
else:
    print("Not a valid triangle")
# Isosceles triangle (5, 5, 8: two sides equal, and 5+5 > 8 so it's valid)
```
