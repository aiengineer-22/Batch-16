# SECTION 2 — How Python Works
---

Before you write a single line of code, you need to understand what actually happens when Python runs your program.

---

### 2.1 — What is a Programming Language?

Your computer only understands one thing: **binary** — sequences of 0s and 1s.  
No human writes 0s and 1s to tell a computer what to do.  
A programming language is a bridge between human-readable instructions and machine instructions.

```
  Human writes:      print("Hello")
  Computer reads:    01001000 01100101 01101100 01101100 01101111

  Something in the middle has to translate.
  For Python, that translator is called the INTERPRETER.
```

---

### 2.2 — Compiled vs Interpreted Languages

There are two approaches to translating code for the machine:

**Compiled languages (e.g. C, C++, Go)**
```
  Source Code  -->  Compiler  -->  Machine Code (executable file)
  
  The entire program is translated BEFORE it runs.
  Very fast at runtime. But you must compile every time you change the code.
```

**Interpreted languages (e.g. Python, JavaScript)**
```
  Source Code  -->  Interpreter  -->  Executed line by line

  Translation happens WHILE the program runs.
  Slightly slower at runtime. But much faster to write and test.
```

Python is interpreted. This is why you can run a single cell in Colab and see the result instantly.  
You do not compile, you do not wait — you write and run immediately.

---

### 2.3 — What Actually Happens When You Run Python Code

Let us trace exactly what happens when you write `print("Hello")` and press Run:

```
STEP 1 — You write source code
         print("Hello")
         This is stored as a plain text file (.py)

STEP 2 — Python reads and checks your code (Lexing & Parsing)
         Python breaks your code into tokens:
         [print] [(] ["Hello"] [)]
         It checks if the grammar is correct.
         If you wrote prnt("Hello") — it stops here with a NameError.

STEP 3 — Bytecode Compilation
         Python translates your code into bytecode.
         Bytecode is an intermediate, lower-level set of instructions.
         It is NOT machine code yet. It is stored in .pyc files.

         print("Hello")  -->  LOAD_GLOBAL 'print'
                              LOAD_CONST  'Hello'
                              CALL_FUNCTION 1

STEP 4 — Python Virtual Machine (PVM) executes bytecode
         The PVM reads bytecode instructions one by one
         and executes them on your actual machine.
         This is the final step — the output appears.
```

**Full picture:**

```
  your_code.py
       |
       v
  [ Python Interpreter ]
       |          |
   Lexer &    Bytecode
   Parser     Compiler
       |          |
       +----+-----+
            |
            v
   [ .pyc bytecode file ]
            |
            v
   [ Python Virtual Machine ]
            |
            v
       OUTPUT on screen
```

### 2.5 — Python is Dynamically Typed

This is one of the most important things to understand about Python.

In statically typed languages (Java, C++), you must declare what type of data a variable holds:

```java
// Java — you must declare the type
int age = 25;
String name = "Ali";
```

In Python, you do not. Python figures out the type automatically at runtime:

```python
# Python — type is inferred automatically
age = 25
name = "Ali"
```

This is called **dynamic typing**. The variable does not have a fixed type — the VALUE assigned to it has a type.

```
  x = 10       -->  x holds an integer
  x = "hello" -->  now x holds a string  (perfectly valid in Python)
  x = 3.14    -->  now x holds a float
```

The same variable can hold different types at different times.  
This makes Python flexible and fast to write, but you must stay aware of what type your variables currently hold.

---

### 2.6 — Python is Object-Oriented (Everything is an Object)

In Python, **everything is an object**.  
Numbers, strings, functions, even the value `True` — they are all objects.

An object is a package that contains:
- A **value** (what it holds)
- A **type** (what kind of thing it is)
- **Methods** (actions it can perform)

```
  "hello"          is a string object
  "hello".upper()  calls the upper() method on that object -> "HELLO"

  42               is an integer object
  42.bit_length()  calls a method on it -> 6 (bits needed to store 42)
```

You will understand objects more deeply when we cover Object Oriented Programming.  
For now, just know: every value in Python comes with built-in capabilities you can call.

### What is a Variable?

A variable is a name that refers to a value stored in memory.  
Think of it as a labeled container — you give it a name, and Python stores the value for you.

```
  age = 25
   |     |
   |     +-- The VALUE being stored (25)
   +-- The NAME you give it (age)

   In memory it looks like this:

   +------------------+
   |  Variable: age   |
   |  Value:    25    |
   |  Type:     int   |
   +------------------+
```

The `=` sign here is the **assignment operator**. It means: take the value on the right and store it under the name on the left.  
It does NOT mean "is equal to" in the mathematical sense.

---

### Rules for Naming Variables

| Valid | Invalid | Reason it is invalid |
|---|---|---|
| `age` | `1age` | Cannot start with a number |
| `user_name` | `user name` | No spaces allowed |
| `total_score` | `total-score` | Hyphen is a minus operator |
| `myData` | `class` | `class` is a reserved Python keyword |
| `_private` | `@value` | Only letters, numbers, underscore allowed |

### Naming Convention: snake_case

Python's standard convention (PEP 8) is **snake_case** — all lowercase, words separated by underscores.

```python
# Poor naming — tells you nothing
x = 150000
d = 30

# Good naming — self-documenting code
monthly_salary = 150000
days_until_deadline = 30
```

Good variable names make your code readable without needing extra comments.

````python
# Creating and using variables
name           = "Arsh"
age            = 24
city           = "Karachi"
is_student     = True
gpa            = 3.85

print(name)
print(age)
print(city)
print(is_student)
print(gpa)
````

````python
# Variables are mutable — their value can change
# This is why they are called VARIables

score = 0
print("Initial score:", score)

score = 10
print("After round 1:", score)

score = score + 5    # Read current value, add 5, store result back
print("After bonus:",  score)

score += 3           # Shorthand for: score = score + 3
print("After round 2:", score)
````

````python
# Python supports multiple assignment in one line
x, y, z = 10, 20, 30
print(x, y, z)

# Assign the same value to multiple variables
a = b = c = 0
print(a, b, c)

# Swap two variables — Python makes this elegant
first  = "Alpha"
second = "Beta"
print("Before swap:", first, second)

first, second = second, first    # Swap in one line
print("After swap: ", first, second)
````

## 6.1 — Integer (int)

Integers are **whole numbers** — no decimal point.  
They can be positive, negative, or zero.  

```
  Positive integers:  1,  42,  1000,  999999
  Negative integers: -1, -42, -1000
  Zero:               0
```

### Integer Operators

| Operator | Symbol | Example | Result |
|---|---|---|---|
| Addition | `+` | `7 + 3` | `10` |
| Subtraction | `-` | `7 - 3` | `4` |
| Multiplication | `*` | `7 * 3` | `21` |
| Division | `/` | `7 / 2` | `3.5` (always returns float) |
| Floor Division | `//` | `7 // 2` | `3` (drops the decimal) |
| Modulo (remainder) | `%` | `7 % 2` | `1` |
| Exponentiation | `**` | `2 ** 8` | `256` |

````python
# Integer examples
num_students  = 45
year          = 2026
temperature   = -5
population    = 1_400_000_000   # Underscores for readability (Python ignores them)

print("Students:",   num_students)
print("Year:",       year)
print("Temperature:",temperature)
print("Population:", population)
````

````python
# Integer arithmetic
a = 17
b = 5

print("Addition:       ", a + b)    # 22
print("Subtraction:    ", a - b)    # 12
print("Multiplication: ", a * b)    # 85
print("Division:       ", a / b)    # 3.4  <-- note: returns float
print("Floor Division: ", a // b)   # 3    <-- drops decimal
print("Modulo:         ", a % b)    # 2    <-- remainder
print("Exponent:       ", a ** b)   # 1419857

print()

# Useful check: is the number even or odd?
number = 42
print(number, "% 2 =", number % 2)   # 0 means even, 1 means odd
````

## 6.2 — Float (float)

Floats are **numbers with a decimal point**.  
They are used whenever precision matters — prices, measurements, percentages, probabilities.

```
  3.14,  -0.5,  99.99,  1.0,  0.001,  1.5e10

  Important distinction:
  5   is an int    (no decimal point)
  5.0 is a float   (has decimal point, even though numerically equal)
```

### Floating Point Precision Warning

Floats are stored in binary (base 2). Some decimal numbers cannot be represented exactly.  
This is a property of all computers, not just Python. You will encounter this in data science work.

````python
# Float examples
pi              = 3.14159
product_price   = 49.99
model_accuracy  = 0.9732    # 97.32% accuracy
learning_rate   = 0.001     # Common value in neural network training

print("Pi:",             pi)
print("Price:",          product_price)
print("Model accuracy:", model_accuracy)
print("Learning rate:",  learning_rate)
````

````python
# Float arithmetic
print("Division always returns float:", 10 / 2)    # 5.0, not 5
print("Int + Float gives float:",       5 + 2.5)   # 7.5

# Scientific notation
very_small = 1.5e-10    # 1.5 x 10^-10
very_large = 2.5e8      # 2.5 x 10^8 = 250,000,000
print("Very small:", very_small)
print("Very large:", very_large)

# Rounding
result = 10 / 3
print("\n10 / 3 =", result)
print("Rounded to 2 decimal places:", round(result, 2))
print("Rounded to 4 decimal places:", round(result, 4))
````

````python
# Floating point precision — understand this, it will save you debugging time
print(0.1 + 0.2)             # You expect 0.3. Look at what Python actually returns.
print(0.1 + 0.2 == 0.3)      # False! This surprises many beginners.

# Why? Because 0.1 and 0.2 cannot be stored exactly in binary floating point.
# The safe way to compare floats:
import math
print(math.isclose(0.1 + 0.2, 0.3))   # True — correct way to compare floats
````

## 6.3 — Complex (complex)

Complex numbers consist of a **real part** and an **imaginary part**.  
In mathematics they are written as `a + bi`.  
In Python, the imaginary unit is written as `j` instead of `i`.

```
  Mathematics:  3 + 4i
  Python:       3 + 4j

  z = 3 + 4j
      |   |
      |   +-- Imaginary part: 4 (multiplied by j, the imaginary unit)
      +-- Real part: 3
```

### Why does this matter for AI and Data Science?

Complex numbers appear in:
- **Signal processing** — analyzing audio, images, time-series data
- **Fourier Transforms** — the mathematical backbone of audio processing and CNNs
- **Quantum computing** — quantum states are described with complex amplitudes
- **Control systems** — stability analysis of models

You will not use them in basic scripts, but they are a native Python type and worth understanding.

````python
# Creating complex numbers
z1 = 3 + 4j
z2 = 1 - 2j
z3 = complex(5, -3)    # Another way: complex(real, imaginary)

print("z1:", z1)
print("z2:", z2)
print("z3:", z3)
print("Type:", type(z1))
````

````python
# Accessing real and imaginary parts
z = 3 + 4j

print("Real part:      ", z.real)        # 3.0
print("Imaginary part: ", z.imag)        # 4.0
print("Conjugate:      ", z.conjugate()) # 3 - 4j  (flips sign of imaginary part)
````

````python
# Arithmetic with complex numbers
z1 = 3 + 4j
z2 = 1 - 2j

print("Addition:       ", z1 + z2)   # (3+1) + (4-2)j = 4 + 2j
print("Subtraction:    ", z1 - z2)   # (3-1) + (4+2)j = 2 + 6j
print("Multiplication: ", z1 * z2)   # Follows (a+bi)(c+di) formula

# Magnitude (absolute value) of a complex number
# For z = a + bj, magnitude = sqrt(a^2 + b^2)
import math
magnitude = abs(z1)
print("Magnitude of", z1, "=", magnitude)   # sqrt(9 + 16) = sqrt(25) = 5.0
````

## 6.4 — String (str)

Strings are **sequences of characters** — any text enclosed in quotes.  
A string can contain letters, numbers, spaces, punctuation, and special characters.

```
  "Hello"         -- double quotes
  'Hello'         -- single quotes (both work equally)
  """Hello"""     -- triple quotes (for multi-line text)

  Important:
  "123"  is a string  (inside quotes)
  123    is an integer (no quotes)
  They look similar but are completely different types.
```

### Strings are Sequences

Each character in a string has an **index** — a position number starting from 0.

```
  word = "Python"

  Index:    0   1   2   3   4   5
  Char:     P   y   t   h   o   n

  Negative: -6  -5  -4  -3  -2  -1   (count from the end)
```

````python
# Creating strings
first_name  = "Hafiza"
last_name   = 'Arshmah'
course      = "Python for AI and Data Science"

# Multi-line string using triple quotes
description = """This course covers:
- Python fundamentals
- Machine Learning
- Deep Learning
- Generative AI"""

print(first_name)
print(last_name)
print(description)
````

````python
# String indexing and slicing
word = "Python"

# Indexing — access a single character
print("First character:  ", word[0])     # P
print("Third character:  ", word[2])     # t
print("Last character:   ", word[-1])    # n  (negative index counts from end)

# Slicing — access a range of characters
# Syntax: string[start : stop]   (stop is NOT included)
print("\nFirst 3 chars:    ", word[0:3])  # Pyt
print("From index 2:     ", word[2:])    # thon
print("Up to index 4:    ", word[:4])    # Pyth
print("Reverse:          ", word[::-1])  # nohtyP  (step -1 means go backwards)
````

````python
# Common string methods
text = "  Python is Amazing  "

print("Original:    |", text, "|")
print("Uppercase:   |", text.upper(),  "|")
print("Lowercase:   |", text.lower(),  "|")
print("Stripped:    |", text.strip(),  "|")
print("Length:      ",  len(text))

print()

sentence = "Data Science is the future of technology"
print("Starts with 'Data'?  ", sentence.startswith("Data"))
print("Contains 'future'?   ", "future" in sentence)
print("Replace 'future':    ", sentence.replace("future", "backbone"))
print("Split into words:    ", sentence.split()[:4], "...")  # shows first 4 words
````

## 6.5 — Boolean (bool)

A Boolean has exactly **two possible values: `True` or `False`**.  
They are the foundation of all decision-making in programming.

```
  is_raining  = True
  is_sunny    = False

  Critical rule:
  True and False MUST start with a capital letter.
  true  or  false  will cause a NameError.
```

### Booleans and Comparisons

Every comparison expression in Python evaluates to a Boolean.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `7 > 3` | `True` |
| `<` | Less than | `2 < 1` | `False` |
| `>=` | Greater or equal | `5 >= 5` | `True` |
| `<=` | Less or equal | `3 <= 2` | `False` |

### Booleans are actually integers under the hood

In Python, `True` is stored as `1` and `False` is stored as `0`.  


````python
# Boolean variables
is_student     = True
has_gpu        = False
model_trained  = True

print("Is student?   ", is_student)
print("Has GPU?      ", has_gpu)
print("Model trained?", model_trained)
````

````python
# Comparison operators always return a boolean
print(10 > 5)      # True
print(10 < 5)      # False
print(10 == 10)    # True
print(10 != 5)     # True
print(10 >= 10)    # True
print(3 <= 2)      # False

print()

# Logical operators combine boolean conditions
age        = 20
has_id     = True

print("Age >= 18 AND has ID:", age >= 18 and has_id)    # Both must be True
print("Age >= 18 OR has ID: ", age >= 18 or has_id)     # At least one must be True
print("NOT has_id:           ", not has_id)              # Flip the value
````

````python
# Booleans are subclass of int — True = 1, False = 0
print("True as integer: ", int(True))    # 1
print("False as integer:", int(False))   # 0

print("True + True:",      True + True)  # 2
print("True * 10:",        True * 10)    # 10

# bool() on different values — what is truthy and what is falsy in Python
print()
print("bool(0):  ",   bool(0))    # False  -- zero is falsy
print("bool(42): ",   bool(42))   # True   -- any non-zero number is truthy
print('bool(""):  ',  bool(""))   # False  -- empty string is falsy
print('bool("hi"):',  bool("hi")) # True   -- non-empty string is truthy
````

## 7.1 — The print() Function in Depth

Before formatting, understand what `print()` can actually do.

```
  print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)

  sep   -- what to put BETWEEN multiple items (default: a single space)
```

## 7.2 — Method 1: % Formatting (Old Style)

This is inherited from C language. Still found in older Python codebases.  
You will encounter it in legacy code, so you should recognize it.

```
  "My name is %s and I am %d years old" % (name, age)

  %s  -- placeholder for a string
  %d  -- placeholder for an integer
  %f  -- placeholder for a float
  %.2f -- float with 2 decimal places
```

````python
# % formatting
name  = "Arqum"
age   = 28
score = 95.678

print("My name is %s" % name)
print("I am %d years old" % age)
print("My score is %.2f" % score)                          # 2 decimal places
print("Name: %s, Age: %d, Score: %.1f" % (name, age, score))  # Multiple values
````

---
## 7.3 — Method 2: .format() Method

Introduced in Python 3 as a cleaner alternative to %.  
Uses curly braces `{}` as placeholders.

```
  "My name is {} and I am {} years old".format(name, age)

  {}         -- positional placeholder
  {0}, {1}   -- indexed placeholder
  {name}     -- named placeholder
  {:.2f}     -- format specifier (2 decimal places)
```

````python
# .format() method
name  = "Arsh"
age   = 23
score = 97.456

# Positional (filled in order)
print("My name is {} and I am {} years old.".format(name, age))

# Indexed (reference by position number)
print("{0} is {1} years old. {0} loves Python.".format(name, age))

# Named placeholders
print("Name: {n}, Age: {a}".format(n=name, a=age))

# Format specifiers
print("Score: {:.2f}".format(score))          # 2 decimal places
print("Score: {:10.2f}".format(score))        # Width 10, 2 decimal places
print("Score: {:>10.2f}".format(score))       # Right-aligned in width 10
print("Score: {:<10.2f}".format(score))       # Left-aligned in width 10
````

## 7.4 — Method 3: f-strings (The Modern Standard)

f-strings (formatted string literals) were introduced in Python 3.6.  
They are the **recommended and most widely used method** today.

```
  f"My name is {name} and I am {age} years old"
   |
   +-- The f prefix tells Python this is an f-string

  Rules:
  - Prefix the string with  f  or  F
  - Put any variable or expression inside  { }
  - Python evaluates what is inside the braces and inserts the result
```

### Why f-strings are preferred
- Shorter and more readable than .format()
- Faster at runtime
- You can embed any Python expression directly inside `{ }`

````python
# f-string basics
name    = "Arsh"
age     = 24
city    = "Karachi"
score   = 88.7456

# Basic variable insertion
print(f"My name is {name} and I am {age} years old.")
print(f"I live in {city}.")
````

````python
# f-strings can contain expressions — not just variable names
a = 10
b = 3

print(f"{a} + {b} = {a + b}")          # Expression inside braces
print(f"{a} * {b} = {a * b}")
print(f"Is a > b? {a > b}")
print(f"a squared = {a ** 2}")
print(f"Name uppercase: {name.upper()}")  # Method call inside braces
````

````python
# f-string formatting specifiers
score   = 88.7456
amount  = 1234567.89

# Decimal places
print(f"Score: {score:.2f}")           # 88.75   (2 decimal places)
print(f"Score: {score:.4f}")           # 88.7456 (4 decimal places)

# Width and alignment
print(f"{'Name':<15} {'Score':>10}")   # Column headers
print(f"{'Fatima':<15} {97.5:>10.2f}")
print(f"{'Mohammed':<15} {88.3:>10.2f}")
print(f"{'Sara':<15} {99.1:>10.2f}")

# Thousands separator
print(f"\nAmount: {amount:,.2f}")       # 1,234,567.89

# Percentage
accuracy = 0.9732
print(f"Model accuracy: {accuracy:.2%}")  # 97.32%
````

````python
# Practical example: formatted report using f-strings
student_name   = "Aisha Al Mansoori"
subject        = "Python Programming"
score          = 94.5
total          = 100
grade          = "A"

print("=" * 40)
print(f"  STUDENT RESULT CARD")
print("=" * 40)
print(f"  Name:     {student_name}")
print(f"  Subject:  {subject}")
print(f"  Score:    {score} / {total}")
print(f"  Percent:  {score/total:.1%}")
print(f"  Grade:    {grade}")
print("=" * 40)
````

# SECTION 8 — The type() Function and Type Conversion
---

### type() — Check the data type of any value

```
  type(25)        -->  <class 'int'>
  type(3.14)      -->  <class 'float'>
  type(3 + 4j)    -->  <class 'complex'>
  type("Hello")   -->  <class 'str'>
  type(True)      -->  <class 'bool'>
```

### isinstance() — A more professional type check

```
  isinstance(25, int)     -->  True
  isinstance("hi", str)   -->  True
  isinstance(3.14, int)   -->  False
```

In production code, `isinstance()` is preferred over `type()` for type checking.

---

### Type Conversion (Casting)

You can explicitly convert one type to another using built-in functions:

| Function | Converts to | Example | Result |
|---|---|---|---|
| `int()` | Integer | `int("25")` | `25` |
| `float()` | Float | `float("3.14")` | `3.14` |
| `str()` | String | `str(100)` | `"100"` |
| `bool()` | Boolean | `bool(0)` | `False` |
| `complex()` | Complex | `complex(3, 4)` | `(3+4j)` |

````python
# type() examples
print(type(42))            # int
print(type(3.14))          # float
print(type(3 + 4j))        # complex
print(type("Hello"))       # str
print(type(True))          # bool
print(type(None))          # NoneType  (special: represents 'no value')
````

````python
# Common beginner confusion
print(type("25"))      # str  -- it is in quotes, so it is text
print(type(25))        # int  -- no quotes
print(type(25.0))      # float -- has decimal point
print(type(True))      # bool
print(type("True"))    # str  -- in quotes, so it is text not boolean
````

````python
# Type conversion examples

# String to Integer
age_text   = "25"           # This is a string
age_number = int(age_text)  # Now it is an integer
print(f"Before: {age_text!r}  type={type(age_text).__name__}")
print(f"After:  {age_number!r}  type={type(age_number).__name__}")

print()

# Integer to Float
score       = 95
score_float = float(score)
print(f"int:   {score!r}   -> float: {score_float!r}")

print()

# Number to String
year     = 2025
year_str = str(year)
print(f"int: {year!r} -> str: {year_str!r}")
print("Can now concatenate: Year " + year_str)
````

````python
# isinstance() — the professional way to check types
values = [42, 3.14, "hello", True, 3+4j]

for val in values:
    print(f"{repr(val):12}  | type: {type(val).__name__:10} | is int? {isinstance(val, int)}")
````

# SECTION 10 — Comments and Best Practices
---

### Comments

Comments are notes in your code. Python ignores them completely when running.  
They exist purely for human readers — yourself in 6 months, and your teammates.

```python
# Single-line comment: starts with #

"""  
Multi-line comment (docstring):
Used to document functions, classes, and modules.
Also used to temporarily disable blocks of code.
"""
```

### When to comment

| Good use | Bad use |
|---|---|
| Explain WHY something is done | Explain WHAT obvious code does |
| Document a non-obvious formula | `x = x + 1  # Add 1 to x` |
| Mark sections in long scripts | Re-stating exactly what the code says |

### PEP 8 — Python's Style Guide

PEP 8 is the official Python style guide. Following it makes your code consistent with every Python developer in the world.

| Rule | Bad | Good |
|---|---|---|
| Variable names | `MyName`, `MYNAME` | `my_name` |
| Spaces around operators | `x=5+3` | `x = 5 + 3` |
| Line length | Very long lines | Keep under 79 characters |
| Blank lines | No spacing | Two blank lines between top-level blocks |

````python
# Example of well-written, properly commented Python code

# ---------------------------------------------------------
# Student Performance Summary
# ---------------------------------------------------------

student_name   = "Khalid Al Nuaimi"
subject        = "Data Science Fundamentals"
marks_obtained = 87
total_marks    = 100

# Calculate percentage and determine grade
percentage = (marks_obtained / total_marks) * 100

# Ternary-style assignment (we will cover if/else properly in the next video)
grade = "A" if percentage >= 90 else "B" if percentage >= 80 else "C"

# Display result
print("-" * 42)
print(f"  Student:    {student_name}")
print(f"  Subject:    {subject}")
print(f"  Score:      {marks_obtained} / {total_marks}")
print(f"  Percentage: {percentage:.1f}%")
print(f"  Grade:      {grade}")
print("-" * 42)
````
